```python
sentence = "AI loves Python"
words = sentence.split()
vocabulary = sorted(set(words))
print(vocabulary)
```

```python
wrod_to_index = {word: index for index, word in enumerate(vocabulary)}
print(wrod_to_index)
```

```python
import numpy as np
for word in vocabulary:
    one_hot_vector = np.zeros(len(vocabulary))
    one_hot_vector[wrod_to_index[word]] = 1
    print(f"One-hot vector for '{word}': {one_hot_vector}")
```

```python
endoded_sentence= []
for word in words:
    one_hot_vector = np.zeros(len(vocabulary))
    one_hot_vector[wrod_to_index[word]] = 1
    endoded_sentence.append(one_hot_vector)
print("Encoded sentence:\n", endoded_sentence)
```

# Lecture 65 – Average Word2Vec (Sentence Embeddings)

```python
import numpy as np
from gensim.models import Word2Vec
```

```python
sentences = [
    ["i", "love", "machine", "learning"],
    ["python", "is", "powerful"],
    ["artificial", "intelligence", "uses", "data"],
    ["machine", "learning", "is", "fun"]
]
```

```python
model = Word2Vec(sentences, vector_size=50, window=2, min_count=1, workers=1,epochs=100)
```

```python
print(model.wv["machine"])
```

```python
def average_word2vec(sentence, model):
    vectors = []

    for word in sentence:
        if word in model.wv:
            vectors.append(model.wv[word])

    if len(vectors) == 0:
        return np.zeros(model.vector_size)

    return np.mean(vectors, axis=0)
```

```python
sentence = ["machine", "learning", "is", "fun"]

embedding = average_word2vec(sentence, model)

print(embedding)
```

```python
print(embedding.shape)
```

```python
sentence1 = ["machine", "learning"]
sentence2 = ["artificial", "intelligence"]

emb1 = average_word2vec(sentence1, model)
emb2 = average_word2vec(sentence2, model)

from numpy.linalg import norm

cosine_similarity = np.dot(emb1, emb2) / (norm(emb1) * norm(emb2))

print(cosine_similarity)
```

```python
from tensorflow.keras.layers import Input
from tensorflow.keras.layers import LSTM

encoder_inputs = Input(shape=(None, 100))

encoder = LSTM(
    64,
    return_state=True
)

encoder_outputs, state_h, state_c = encoder(encoder_inputs)

print(state_h.shape)
```

```python
decoder_inputs = Input(shape=(None, 100))

decoder = LSTM(
    64,
    return_sequences=True,
    return_state=True
)

decoder_outputs, _, _ = decoder(
    decoder_inputs,
    initial_state=[state_h, state_c]
)
```
