```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text = """
LangChain helps build AI applications. It works with documents, prompts, memory, and retrievers.
"""

splitter = RecursiveCharacterTextSplitter(
    chunk_size=50,
    chunk_overlap=10
)

docs = splitter.split_text(text)
print(docs)
```

```python
conversation=[]

while True:
    user_input = input("User: ")
    if user_input.lower() in ["exit", "quit"]:
        print("Exiting the conversation.")
        break

    conversation.append({"role": "user", "content": user_input})

    response = chain.invoke({"topic": user_input, "level": "beginner"})
    print(f"AI: {response}")

    conversation.append({"role": "assistant", "content": response})
    print("Conversation history:", conversation)
    print("Assistant:", response)
    
```

```python
import numpy as np

documents = {
    "AI": np.array([0.9, 0.8, 0.2]),
    "Machine Learning": np.array([0.85, 0.82, 0.25]),
    "Football": np.array([0.1, 0.2, 0.95])
}

for name, vector in documents.items():
    print(name, ":", vector)
```

```python
query = np.array([0.88, 0.79, 0.22])
for name, vector in documents.items():
    distance = np.linalg.norm(query - vector)
    print(f"Distance from query to {name}: {distance}")
```

```python
from sklearn.metrics.pairwise import cosine_similarity
query= np.array([[0.88, 0.79, 0.22]])
for name, vector in documents.items():
    vector_reshaped = vector.reshape(1, -1)
    similarity = cosine_similarity(query, vector_reshaped)[0][0]
    print(f"Cosine similarity from query to {name}: {similarity}")
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
from langchain_community.vectorstores import FAISS

faiss_db = FAISS.from_texts(
    documents,
    embedding_model
)
```

```python
import streamlit as st
from langchain_ollama import OllamaLLM

st.set_page_config(page_title="Simple GenAI App", page_icon="🤖")

st.title("Simple GenAI App Using Ollama")
st.write("Type your prompt below and generate a response from a local model.")

user_input = st.text_area("Enter your prompt here:")

if st.button("Generate Response"):
    if user_input.strip():
        llm = OllamaLLM(model="llama3")
        response = llm.invoke(user_input)
        st.subheader("Response")
        st.write(response)
    else:
        st.warning("Please enter a prompt first.")
```
