# stemming

```python
# playing   play  played player    
```

```python
from nltk.stem import  PorterStemmer
stemmer = PorterStemmer()
```

```python
words = ["playing", "play", "plays", "player","played"]

for word in words:
    print(word, "->", stemmer.stem(word))
```

```python
words=["studies","studying", "study"]
for word in words:
    print(word, "->", stemmer.stem(word))
```

```python
sentence = "The students were studying machine learning."

tokens = sentence.split()

for token in tokens:
    print(token, "→", stemmer.stem(token))
```

```python
more_words = [
    "running",
    "runner",
    "runs",
    "easily",
    "fairly"
]

for word in more_words:
    print(word, "→", stemmer.stem(word))
```

# Lemmatization 


```python
# studies  -> studi  
```

```python
from nltk.stem import WordNetLemmatizer
nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
words=["playing", "play","played"]
for word in words:
    print(word, "->",lemmatizer.lemmatize(word, pos="v"))
```

```python
words = [
    "studies",
    "studying",
    "study"
]

for word in words:
    print(word, "→", lemmatizer.lemmatize(word, pos="v"))
```

```python
words = [
    "cars",
    "children",
    "mice",
    "geese"
]

for word in words:
    print(word, "→", lemmatizer.lemmatize(word, pos="n"))
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

words = [
    "studies",
    "running",
    "better",
    "geese"
]

print("Word\t\tStem\t\tLemma")

for word in words:
    stem = stemmer.stem(word)
    lemma = lemmatizer.lemmatize(word, pos="v")
    print(f"{word}\t\t{stem}\t\t{lemma}")
```

```python
# mini project 1 - news article analyzer 
import nltk 
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag, ne_chunk

nltk.download("punkt")
nltk.download("stopwords")
nltk.download("averaged_perceptron_tagger")
nltk.download("maxent_ne_chunker")
nltk.download("words")
```

```python
text = """
OpenAI has recently released a new version of its language model, GPT-4. The model has shown significant improvements in natural language understanding and generation. Many developers and researchers are excited about the potential applications of GPT-4 in various fields, including healthcare, education, and customer service.
The release of GPT-4 has sparked discussions about the ethical implications of advanced AI models. Experts are emphasizing the importance of responsible AI development and deployment. OpenAI has also introduced new safety measures to mitigate potential risks associated with the use of GPT-4.
Researchers are exploring ways to leverage GPT-4 for tasks such as medical diagnosis, content creation, and language translation. The model's ability to understand context and generate coherent responses has opened up new possibilities for human-computer interaction. As AI technology continues to evolve, the impact of models like GPT-4 on society and industry is expected to grow.
"""
```

```python
text = text.lower()
print("Lowercase Text:\n", text)
```

```python
translator = str.maketrans("", "", string.punctuation)
text = text.translate(translator)
print("Text after removing punctuation:\n", text)
```

```python
tokens=word_tokenize(text)
print("Tokens:\n", tokens)
print("Number of tokens:", len(tokens))
```

```python
stop_words = set(stopwords.words("english"))
filtered_tokens = [word for word in tokens if word not in stop_words]
print("Filtered Tokens:\n", filtered_tokens)
```

```python
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]
print("Lemmatized Tokens:\n", lemmatized_tokens)
```

```python
tags = ne_chunk(pos_tag(lemmatized_tokens))
print("Named Entities:\n", tags)
```

```python
entities = ne_chunk(tags)
print("Named Entities:\n", entities)
print("Named Entities:", [entity[0] for entity in entities])


```
